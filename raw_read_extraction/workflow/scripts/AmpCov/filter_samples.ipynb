```python
import pandas as pd
```

```python
dt_all_samples = pd.read_csv('samples.tsv', sep='\t', names=['sample_id', 'batch', 'length', 'primer_protocol'])
dt_my_samples = pd.read_csv('../../results_haplo_SarsCov2_21_22/timeline.tsv', sep='\t')
```

```python
locations_of_interest = ["Zürich (ZH)", "Lugano (TI)", "Genève (GE)", "Chur (GR)", "Laupen (BE)", "Altenrhein (SG)"]
```

```python

dt_my_samples = dt_my_samples[(dt_my_samples['date'] >= '2021-11-21') & (dt_my_samples['date'] <= '2022-02-28') & (dt_my_samples['location'].isin(locations_of_interest))]
```

```python
dt_target_samples = dt_all_samples[dt_all_samples['sample_id'].isin(dt_my_samples['sample'])]
```
